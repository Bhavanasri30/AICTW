```python
%pip install -q langchain-community langchain-text-splitters langchain-huggingface
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain_community.document_loaders import PyPDFLoader


# 1. Load the PDF
# Place company_policy.pdf in the same folder as your notebook or script.
loader = PyPDFLoader("company_policy.pdf")
documents = loader.load()

print(f"Loaded {len(documents)} pages.")


# 2. Split the documents into smaller chunks
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_documents(documents)

print(f"Created {len(chunks)} chunks.")


# 3. Load the embedding model
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)


# 4. Generate embeddings and store them in FAISS
vectorstore = FAISS.from_documents(
    documents=chunks,
    embedding=embeddings
)


# 5. Create a retriever
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 4}
)


# 6. Define the query
query = "What is the leave policy?"


# 7. Retrieve relevant chunks
retrieved_documents = retriever.invoke(query)


# 8. Display the retrieved content
for index, document in enumerate(retrieved_documents, start=1):
    page = document.metadata.get("page")
    page_number = page + 1 if page is not None else "Unknown"

    print(f"\n--- Result {index} | Page {page_number} ---")
    print(document.page_content)
```
