## Univariate analysis 

1. what is the age distribution of customers ? 

```python
plt.figure(figsize=(8,6))
plt.hist(df['age'],bins = 15 , edgecolor = 'black',color = 'pink')
plt.xlabel('Age')
plt.ylabel('count')
plt.title('Age Distribution of customers')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------
# 1. Remove missing age values
# ------------------------------------------

ages = df['age'].dropna()


# ------------------------------------------
# 2. Create age groups
# ------------------------------------------

start = int(ages.min() // 10 * 10)
end = int(ages.max() // 10 * 10 + 10)

bins = np.arange(start, end + 1, 10)

counts, edges = np.histogram(
    ages,
    bins=bins
)


# ------------------------------------------
# 3. Create age-group labels
# ------------------------------------------

labels = [
    f'{int(edges[i])}-{int(edges[i+1]-1)}'
    for i in range(len(counts))
]


# ------------------------------------------
# 4. Prepare positions for 3D bars
# ------------------------------------------

xpos = np.arange(len(counts)) * 1.4

ypos = np.zeros(len(counts))

zpos = np.zeros(len(counts))

# Bar width
dx = np.ones(len(counts)) * 0.8

# Bar depth
dy = np.ones(len(counts)) * 0.7

# Bar height = number of customers
dz = counts


# ------------------------------------------
# 5. Create 3D figure
# ------------------------------------------

fig = plt.figure(figsize=(14, 8))

# Outer background
fig.patch.set_facecolor('#EAF2F8')

ax = fig.add_subplot(
    111,
    projection='3d'
)

# Graph background
ax.set_facecolor('#F8FBFD')


# ------------------------------------------
# 6. Create 3D bars
# ------------------------------------------

ax.bar3d(
    xpos,
    ypos,
    zpos,
    dx,
    dy,
    dz,
    color='#4F86C6',
    edgecolor='black',
    linewidth=0.6,
    alpha=0.9,
    shade=True
)


# ------------------------------------------
# 7. Age-group labels
# ------------------------------------------

ax.set_xticks(
    xpos + dx[0] / 2
)

ax.set_xticklabels(
    labels,
    fontsize=9,
    rotation=25,
    ha='right'
)


# ------------------------------------------
# 8. Hide unnecessary depth axis
# ------------------------------------------

ax.set_yticks([])


# ------------------------------------------
# 9. Add count above every bar
# ------------------------------------------

for x, count in zip(xpos, counts):

    ax.text(
        x + 0.4,
        0.35,
        count + max(counts) * 0.02,
        f'{count}',
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold'
    )


# ------------------------------------------
# 10. Axis labels
# ------------------------------------------

ax.set_xlabel(
    'Age Group (Years)',
    fontsize=12,
    fontweight='bold',
    labelpad=20
)

ax.set_zlabel(
    'Number of Customers',
    fontsize=12,
    fontweight='bold',
    labelpad=12
)


# ------------------------------------------
# 11. Title
# ------------------------------------------

ax.set_title(
    '3D Distribution of Customers by Age Group',
    fontsize=17,
    fontweight='bold',
    pad=25
)


# ------------------------------------------
# 12. Best 3D viewing angle
# ------------------------------------------

ax.view_init(
    elev=25,
    azim=-60
)


# ------------------------------------------
# 13. Z-axis limit
# ------------------------------------------

ax.set_zlim(
    0,
    max(counts) * 1.15
)


# ------------------------------------------
# 14. Grid
# ------------------------------------------

ax.grid(True)


# ------------------------------------------
# 15. Adjust spacing
# ------------------------------------------

plt.subplots_adjust(
    left=0.05,
    right=0.95,
    bottom=0.15,
    top=0.90
)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

ages = df['age'].dropna()

start = int(ages.min() // 10 * 10)
end = int(ages.max() // 10 * 10 + 10)
bins = np.arange(start, end + 1, 10)

counts, edges = np.histogram(ages, bins=bins)
labels = [f'{int(edges[i])}–{int(edges[i+1] - 1)}'
          for i in range(len(counts))]

fig = plt.figure(figsize=(11, 7))
ax = fig.add_subplot(111, projection='3d')

x = np.arange(len(labels))
colors = plt.cm.Blues(np.linspace(0.45, 0.85, len(labels)))

ax.bar3d(x, np.zeros(len(x)), np.zeros(len(x)),
         0.65, 0.5, counts,
         color=colors, edgecolor='white', shade=True)

ax.set_xticks(x + 0.325)
ax.set_xticklabels(labels, rotation=20, ha='right')
ax.set_yticks([])  # Depth does not represent another variable
ax.set_zlabel('Number of customers', labelpad=10)
ax.set_xlabel('Age group (years)', labelpad=12)
ax.set_title('Customers by Age Group', fontsize=16, weight='bold', pad=20)
ax.view_init(elev=22, azim=-60)

plt.tight_layout()
plt.show()
```

## observation: 

- Most customers are between 30 and 60 years old.
- The largest group is around 30–40 years old.
- The number of customers generally decreases as age increases beyond 40.
- Very few customers are under 25 or over 65.

This chart shows how many customers are in each age group. It does not show which age group is more likely to give a “yes” response; for that, compare age groups against the response column.

11. which job category have the highest postitive response rate ?

```python
job_response = pd.crosstab(
    df['job'], df['response'], normalize='index'
) * 100

job_response['yes'].sort_values().plot(
    kind='barh', figsize=(8, 6)
)

plt.xlabel('Response Rate (%)')
plt.ylabel('Job category')
plt.title('Response Rate by Job Category')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# 1. Calculate YES response rate
# --------------------------------------------------

job_response = pd.crosstab(
    df['job'],
    df['response'],
    normalize='index'
) * 100

job_yes = job_response['yes'].sort_values()

jobs = job_yes.index
rates = job_yes.values


# --------------------------------------------------
# 2. Give MORE SPACE between job categories
# --------------------------------------------------

ypos = np.arange(len(jobs)) * 2.0

xpos = np.zeros(len(jobs))
zpos = np.zeros(len(jobs))

dx = rates

# Smaller bar thickness
dy = np.ones(len(jobs)) * 0.7
dz = np.ones(len(jobs)) * 0.5


# --------------------------------------------------
# 3. Create large 3D figure
# --------------------------------------------------

fig = plt.figure(figsize=(18, 12))

fig.patch.set_facecolor('#EAF2F8')

ax = fig.add_subplot(111, projection='3d')

ax.set_facecolor('#F8FBFD')


# --------------------------------------------------
# 4. Create 3D bars
# --------------------------------------------------

ax.bar3d(
    xpos,
    ypos,
    zpos,
    dx,
    dy,
    dz,
    color='mediumseagreen',
    edgecolor='black',
    linewidth=0.6,
    alpha=0.9,
    shade=True
)


# --------------------------------------------------
# 5. Job names
# --------------------------------------------------

ax.set_yticks(ypos + 0.35)

ax.set_yticklabels(
    jobs,
    fontsize=9,
    fontweight='bold'
)


# --------------------------------------------------
# 6. Hide depth axis
# --------------------------------------------------

ax.set_zticks([])


# --------------------------------------------------
# 7. Add percentages
# --------------------------------------------------

for i, rate in enumerate(rates):

    ax.text(
        rate + 0.3,
        ypos[i] + 0.35,
        0.3,
        f'{rate:.1f}%',
        fontsize=9,
        fontweight='bold',
        ha='left',
        va='center'
    )


# --------------------------------------------------
# 8. Labels
# --------------------------------------------------

ax.set_xlabel(
    'YES Response Rate (%)',
    fontsize=13,
    fontweight='bold',
    labelpad=15
)

ax.set_ylabel(
    'Job Category',
    fontsize=13,
    fontweight='bold',
    labelpad=20
)


# --------------------------------------------------
# 9. Title
# --------------------------------------------------

ax.set_title(
    '3D Customer Response Rate by Job Category',
    fontsize=18,
    fontweight='bold',
    pad=25
)


# --------------------------------------------------
# 10. Axis limits
# --------------------------------------------------

ax.set_xlim(0, max(rates) + 5)

ax.set_ylim(
    -0.5,
    max(ypos) + 2
)


# --------------------------------------------------
# 11. Better viewing angle
# --------------------------------------------------

ax.view_init(
    elev=18,
    azim=-60
)


# --------------------------------------------------
# 12. Grid
# --------------------------------------------------

ax.grid(True)


# --------------------------------------------------
# 13. Give more space for names
# --------------------------------------------------

plt.subplots_adjust(
    left=0.18,
    right=0.95,
    bottom=0.10,
    top=0.90
)

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# --------------------------------------------------
# 1. Calculate monthly response rate
# --------------------------------------------------

month_response = pd.crosstab(
    df['month_name'],
    df['response'],
    normalize='index'
) * 100


# --------------------------------------------------
# 2. Arrange months in correct order
# --------------------------------------------------

month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

monthly_response = month_response.reindex(month_order)


# --------------------------------------------------
# 3. Remove months that don't exist in dataset
# --------------------------------------------------

monthly_response = monthly_response.dropna(subset=['yes'])


# --------------------------------------------------
# 4. Prepare data
# --------------------------------------------------

months = monthly_response.index
response_rate = monthly_response['yes'].values

x = np.arange(len(months))

# Depth position
y = np.zeros(len(months))

# --------------------------------------------------
# 5. Create 3D figure
# --------------------------------------------------

fig = plt.figure(figsize=(16, 9))

fig.patch.set_facecolor('#EAF2F8')

ax = fig.add_subplot(111, projection='3d')

ax.set_facecolor('#F8FBFD')


# --------------------------------------------------
# 6. Create 3D line
# --------------------------------------------------

ax.plot(
    x,
    y,
    response_rate,
    marker='o',
    markersize=9,
    linewidth=3,
    color='royalblue',
    markerfacecolor='orange',
    markeredgecolor='black',
    label='YES Response Rate'
)


# --------------------------------------------------
# 7. Add vertical lines for 3D effect
# --------------------------------------------------

for i, rate in enumerate(response_rate):

    ax.plot(
        [x[i], x[i]],
        [0, 0],
        [0, rate],
        linestyle='--',
        linewidth=1.2,
        color='gray',
        alpha=0.6
    )


# --------------------------------------------------
# 8. Add percentage values
# --------------------------------------------------

for i, rate in enumerate(response_rate):

    ax.text(
        x[i],
        0,
        rate + 0.5,
        f'{rate:.1f}%',
        fontsize=9,
        fontweight='bold',
        ha='center'
    )


# --------------------------------------------------
# 9. Month labels
# --------------------------------------------------

ax.set_xticks(x)

ax.set_xticklabels(
    months,
    rotation=35,
    ha='right',
    fontsize=9
)


# --------------------------------------------------
# 10. Hide unnecessary depth-axis labels
# --------------------------------------------------

ax.set_yticks([])


# --------------------------------------------------
# 11. Axis labels
# --------------------------------------------------

ax.set_xlabel(
    'Month',
    fontsize=13,
    fontweight='bold',
    labelpad=22
)

ax.set_zlabel(
    'YES Response Rate (%)',
    fontsize=13,
    fontweight='bold',
    labelpad=12
)


# --------------------------------------------------
# 12. Professional title
# --------------------------------------------------

ax.set_title(
    'Monthly Customer Response Rate Analysis',
    fontsize=18,
    fontweight='bold',
    pad=25
)


# --------------------------------------------------
# 13. Adjust Z-axis
# --------------------------------------------------

ax.set_zlim(
    0,
    max(response_rate) + 5
)


# --------------------------------------------------
# 14. Best 3D viewing angle
# --------------------------------------------------

ax.view_init(
    elev=25,
    azim=-60
)


# --------------------------------------------------
# 15. Grid
# --------------------------------------------------

ax.grid(True)


# --------------------------------------------------
# 16. Legend
# --------------------------------------------------

ax.legend(
    loc='upper left',
    fontsize=10
)


# --------------------------------------------------
# 17. Adjust layout
# --------------------------------------------------

plt.subplots_adjust(
    left=0.05,
    right=0.95,
    bottom=0.15,
    top=0.90
)

plt.show()
```

13. How does response vary when education and martial status are considered together?

```python
education_marital = pd.crosstab(
    [df['education'], df['marital']],
    df['response'],
    normalize='index'
) * 100

education_marital.plot(
    kind='bar',
    figsize=(14, 7),
    edgecolor='black'
)

plt.xlabel('Education and Marital Status')
plt.ylabel('Response Rate (%)')
plt.title('Response Rate by Education and Marital Status')

plt.xticks(rotation=45, ha='right')

plt.legend(
    title='Response',
    title_fontsize=13,
    fontsize=11
)

plt.grid(axis='y', linestyle='-.')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------
# 1. Calculate YES response rate
# ------------------------------------------------

education_marital = pd.crosstab(
    [df['education'], df['marital']],
    df['response'],
    normalize='index'
) * 100

# Take only YES response percentage
response_rate = education_marital['yes'].unstack()

# ------------------------------------------------
# 2. Get education and marital categories
# ------------------------------------------------

education = response_rate.index
marital = response_rate.columns

# ------------------------------------------------
# 3. Create positions for 3D bars
# ------------------------------------------------

xpos, ypos = np.meshgrid(
    np.arange(len(education)),
    np.arange(len(marital)),
    indexing='ij'
)

xpos = xpos.flatten()
ypos = ypos.flatten()

# Bars start from 0
zpos = np.zeros_like(xpos)

# Width and depth of bars
dx = np.ones_like(xpos) * 0.5
dy = np.ones_like(ypos) * 0.5

# Height = YES response percentage
dz = response_rate.values.flatten()

# ------------------------------------------------
# 4. Create figure
# ------------------------------------------------

fig = plt.figure(figsize=(16, 10))

# Background
fig.patch.set_facecolor('#EAF2F8')

ax = fig.add_subplot(111, projection='3d')

ax.set_facecolor('#F8FBFD')

# ------------------------------------------------
# 5. Create 3D bars
# ------------------------------------------------

bars = ax.bar3d(
    xpos,
    ypos,
    zpos,
    dx,
    dy,
    dz,
    color='mediumseagreen',
    edgecolor='black',
    linewidth=0.6,
    alpha=0.9,
    shade=True
)

# ------------------------------------------------
# 6. X-axis → Education
# ------------------------------------------------

ax.set_xticks(np.arange(len(education)) + 0.25)

ax.set_xticklabels(
    education,
    fontsize=10,
    rotation=20,
    ha='right'
)

# ------------------------------------------------
# 7. Y-axis → Marital Status
# ------------------------------------------------

ax.set_yticks(np.arange(len(marital)) + 0.25)

ax.set_yticklabels(
    marital,
    fontsize=10
)

# ------------------------------------------------
# 8. Axis labels
# ------------------------------------------------

ax.set_xlabel(
    'Education',
    fontsize=13,
    fontweight='bold',
    labelpad=18
)

ax.set_ylabel(
    'Marital Status',
    fontsize=13,
    fontweight='bold',
    labelpad=18
)

ax.set_zlabel(
    'YES Response Rate (%)',
    fontsize=13,
    fontweight='bold',
    labelpad=12
)

# ------------------------------------------------
# 9. Title
# ------------------------------------------------

ax.set_title(
    '3D Customer Response Rate by Education & Marital Status',
    fontsize=17,
    fontweight='bold',
    pad=25
)

# ------------------------------------------------
# 10. Add percentage values above bars
# ------------------------------------------------

for x, y, height in zip(xpos, ypos, dz):

    if not np.isnan(height):

        ax.text(
            x + 0.25,
            y + 0.25,
            height + 1,
            f'{height:.1f}%',
            ha='center',
            va='bottom',
            fontsize=8,
            fontweight='bold'
        )

# ------------------------------------------------
# 11. Z-axis range
# ------------------------------------------------

ax.set_zlim(
    0,
    max(30, np.nanmax(dz) + 10)
)

# ------------------------------------------------
# 12. Professional viewing angle
# ------------------------------------------------

ax.view_init(
    elev=25,
    azim=-55
)

# Grid
ax.grid(True)

# Adjust spacing
plt.subplots_adjust(
    left=0.05,
    right=0.95,
    bottom=0.12,
    top=0.90
)

plt.show()
```
