## data description

Here’s a description of each column in your dataset:

| Column       | Description                                                                                                                           |
| ------------ | ------------------------------------------------------------------------------------------------------------------------------------- |
| `customerid` | Unique ID assigned to each customer.                                                                                                  |
| `age`        | Customer’s age in years.                                                                                                              |
| `salary`     | Customer’s salary or estimated income.                                                                                                |
| `balance`    | Customer’s bank account balance.                                                                                                      |
| `marital`    | Customer’s marital status.                                                                                                            |
| `jobedu`     | Customer’s job and education level together, such as `technician,secondary`.                                                          |
| `targeted`   | Whether the customer was marked for targeting in the dataset.                                                                         |
| `default`    | Whether the customer has a credit default recorded.                                                                                   |
| `housing`    | Whether the customer has a housing loan.                                                                                              |
| `loan`       | Whether the customer has a personal loan.                                                                                             |
| `contact`    | Method used to contact the customer, such as cellular or telephone.                                                                   |
| `day`        | Day of the month when the latest contact happened.                                                                                    |
| `month`      | Month and year of the latest contact, such as `may, 2017`.                                                                            |
| `duration`   | Length of the latest call. Your data contains values in both seconds and minutes.                                                     |
| `campaign`   | Number of contacts during the current campaign, including the latest contact.                                                         |
| `pdays`      | Days since a contact in an earlier campaign. `-1` appears to indicate no earlier contact.                                             |
| `previous`   | Number of contacts before the current campaign.                                                                                       |
| `poutcome`   | Outcome of the previous campaign’s contact.                                                                                           |
| `response`   | Whether the customer accepted the current offer (`yes` or `no`). This is the **target column** if you’re building a prediction model. |

```python
df['job'] = df.jobedu.apply(lambda x : x.split(',')[0])
df.head(2) 
```

## Invalid values converting duration column in minutes from sec

```python
df['duration'] = df['duration'].apply(lambda x : float(x.replace('sec', '') .strip())/60  # to convert the datatype of duration column to integer
                                     if 'sec' in x 
                                     else float(x.replace('min', '') .strip()))   # to convert the datatype of duration column to integer 
df.head(2)
```

## Univariate analysis 

1. what is the age distribution of customers ? 

```python
plt.figure(figsize=(8,6))
plt.hist(df['age'],bins = 15 , edgecolor = 'black',color = 'pink')
plt.xlabel('Age')
plt.ylabel('count')
plt.title('Age Distribution of customers')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

ages = df['age'].dropna()

# Create age groups: 20–29, 30–39, etc.
start = int(ages.min() // 10 * 10)
end = int(ages.max() // 10 * 10 + 10)
bins = np.arange(start, end + 1, 10)

counts, edges = np.histogram(ages, bins=bins)
labels = [f'{int(edges[i])}–{int(edges[i+1]-1)}'
          for i in range(len(counts))]

plt.figure(figsize=(10, 6))
bars = plt.bar(labels, counts, color='#4F86C6', edgecolor='white', width=0.7)

plt.bar_label(bars, padding=4, fontsize=10)
plt.xlabel('Age group (years)')
plt.ylabel('Number of customers')
plt.title('Customers by Age Group', fontsize=16, weight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.25)
plt.gca().set_axisbelow(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

ages = df['age'].dropna()

start = int(ages.min() // 10 * 10)
end = int(ages.max() // 10 * 10 + 10)
bins = np.arange(start, end + 1, 10)

counts, edges = np.histogram(ages, bins=bins)
labels = [f'{int(edges[i])}–{int(edges[i+1] - 1)}'
          for i in range(len(counts))]

fig = plt.figure(figsize=(11, 7))
ax = fig.add_subplot(111, projection='3d')

x = np.arange(len(labels))
colors = plt.cm.Blues(np.linspace(0.45, 0.85, len(labels)))

ax.bar3d(x, np.zeros(len(x)), np.zeros(len(x)),
         0.65, 0.5, counts,
         color=colors, edgecolor='white', shade=True)

ax.set_xticks(x + 0.325)
ax.set_xticklabels(labels, rotation=20, ha='right')
ax.set_yticks([])  # Depth does not represent another variable
ax.set_zlabel('Number of customers', labelpad=10)
ax.set_xlabel('Age group (years)', labelpad=12)
ax.set_title('Customers by Age Group', fontsize=16, weight='bold', pad=20)
ax.view_init(elev=22, azim=-60)

plt.tight_layout()
plt.show()
```

## observation: 

- Most customers are between 30 and 60 years old.
- The largest group is around 30–40 years old.
- The number of customers generally decreases as age increases beyond 40.
- Very few customers are under 25 or over 65.

This chart shows how many customers are in each age group. It does not show which age group is more likely to give a “yes” response; for that, compare age groups against the response column.

```python
import matplotlib.pyplot as plt

marital_counts = df['marital'].value_counts()

colors = ['#4F86C6', '#F4A261', '#81C7A8', '#B39DDB']

fig, ax = plt.subplots(figsize=(8, 6))

wedges, texts, autotexts = ax.pie(
    marital_counts,
    labels=marital_counts.index.str.title(),
    autopct=lambda pct: f'{pct:.1f}%\n({round(pct * marital_counts.sum() / 100):,})',
    startangle=90,
    colors=colors[:len(marital_counts)],
    wedgeprops={'edgecolor': 'white', 'linewidth': 2},
    textprops={'fontsize': 11},
    pctdistance=0.7
)

ax.set_title('Customers by Marital Status', fontsize=16, weight='bold')
ax.axis('equal')  # Keeps the pie circular
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

marital_counts = df['marital'].value_counts()
total = marital_counts.sum()

labels = marital_counts.index.str.title()
counts = marital_counts.values
x = np.arange(len(labels))

colors = ['#5B8DEF', '#FFB86B', '#77C9A3', '#B69BE0']

fig = plt.figure(figsize=(9, 6))
ax = fig.add_subplot(111, projection='3d')

ax.bar3d(
    x, np.zeros(len(x)), np.zeros(len(x)),
    0.6, 0.5, counts,
    color=colors[:len(x)],
    edgecolor='white',
    shade=True
)

for i, count in enumerate(counts):
    ax.text(
        x[i] + 0.3, 0.25, count + counts.max() * 0.03,
        f'{count:,}\n({count / total:.1%})',
        ha='center', fontsize=10
    )

ax.set_xticks(x + 0.3)
ax.set_xticklabels(labels)
ax.set_yticks([])  # Depth is only for appearance
ax.set_zlabel('Number of customers')
ax.set_title('Customers by Marital Status', fontsize=15, weight='bold', pad=20)
ax.set_zlim(0, counts.max() * 1.25)
ax.view_init(elev=20, azim=-60)

plt.tight_layout()
plt.show()
```

9. Does the number of customers who responded differ between customers with and without personal loans?
