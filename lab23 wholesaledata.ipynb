```python
# reading the data and looking at the first five rows of the 

data = pd.read_csv(r"C:\Users\BHAVANASRI\OneDrive\AICW\Lab 23 Wholesale customers data.csv")
data.head()
```

```python
data.describe()
```

```python
# strandadizing the data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data)


# statitics of scaled data 
pd.DataFrame(data_scaled).describe()
```

```python
# defining the Kmeans functrion with initilization as kmeans ++

kMeans = KMeans(n_clusters=2 , init = 'k-means++')

# fitting the k means algorithm on scaled data 
kMeans.fit(data_scaled)
```

```python
##inertia on the fitted data
kMeans.inertia_
```

```python
# fitting multiple k-means algorithms and storing the values in an empty list
SSE = []
for cluster in range(1,20):
    kmeans = KMeans(n_clusters = cluster, init='k-means++')
    kmeans.fit(data_scaled)
    SSE.append(kmeans.inertia_)

# converting the results into a dataframe and plotting them
frame = pd.DataFrame({'Cluster':range(1,20), 'SSE':SSE})
plt.figure(figsize=(12,6))
plt.plot(frame['Cluster'], frame['SSE'], marker='o')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
```

```python
# Kmeans with 5 clusters and random state as 42
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42)
kmeans.fit(data_scaled)
pred = kmeans.predict(data_scaled)
```
