## Encoding

| Term | Definition | Where it is used | Example |
|---|---|---|---|
| **Label encoding** | Replaces each category with a number. | Usually for a **categorical target** in classification. | `No → 0`, `Yes → 1` |
| **Target encoding** | Replaces each input category with a value calculated from the target, often the average target value for that category. | For an **input column with many categories**, such as neighbourhood. | A neighbourhood becomes the average house price of homes there in the training data. |
| **One-hot encoding** | Creates a separate `0`/`1` column for each category. | For **input categories with no natural order**, such as garage type. | `Attchd → [1, 0]`, `Detchd → [0, 1]` |

| Type | Meaning | Example from house data | Suitable encoding |
|---|---|---|---|
| **Nominal categorical** | Categories have **no meaningful order** | `GarageType`: Attchd, Detchd, BuiltIn | One-hot encoding |
| **Ordinal categorical** | Categories have a **meaningful order** | `KitchenQual`: Poor < Fair < Typical < Good < Excellent | Ordinal encoding |

```python
# Select the categorical columns that have a meaningful order
ordinal_candidates = [
    'ExterQual', 'ExterCond', 'BsmtQual', 'BsmtCond',
    'HeatingQC', 'KitchenQual', 'FireplaceQu',
    'GarageQual', 'GarageCond', 'PoolQC'
]

ordinal_columns = [
    col for col in ordinal_candidates if col in categorical_columns
]

# The remaining categorical columns are treated as nominal
nominal_columns = [
    col for col in categorical_columns if col not in ordinal_columns
]

print("Ordinal columns:")
print(ordinal_columns)

print("\nNominal columns:")
print(nominal_columns)
```
